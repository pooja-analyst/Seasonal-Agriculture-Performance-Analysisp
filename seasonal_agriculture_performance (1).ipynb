```python
from scipy.stats import f_oneway
kharif_yield = df_clean[df_clean["Season"] == "Kharif"]["Yield_Tonnes_Ha"]
rabi_yield = df_clean[df_clean["Season"] == "Rabi"]["Yield_Tonnes_Ha"]
zaid_yield = df_clean[df_clean["Season"] == "Zaid"]["Yield_Tonnes_Ha"]

f_stat, p_value = f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)
print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
from scipy.stats import f_oneway
kharif_profit = df_clean[df_clean["Season"] == "Kharif"]["Profit_INR"]
rabi_profit = df_clean[df_clean["Season"] == "Rabi"]["Profit_INR"]
zaid_profit = df_clean[df_clean["Season"] == "Zaid"]["Profit_INR"]

f_stat_profit, p_value_profit = f_oneway(
    kharif_profit,
    rabi_profit,
    zaid_profit
)
print("F-statistic:", f_stat_profit)
print("P-value:", p_value_profit)
```

```python
crop_yield_groups = [
    group["Yield_Tonnes_Ha"].values
    for _, group in df_clean.groupby("Crop")
]

f_stat_crop, p_value_crop = f_oneway(*crop_yield_groups)
print("F-statistic:", f_stat_crop)
print("P-value:", p_value_crop)
```

```python
irrigation_yield_groups = [
    group["Yield_Tonnes_Ha"].values
    for _, group in df_clean.groupby("Irrigation_Method")
]
f_stat_irrigation, p_value_irrigation = f_oneway(
    *irrigation_yield_groups
)
print("F-statistic:", f_stat_irrigation)
print("P-value:", p_value_irrigation)
```

```python
from scipy.stats import f_oneway

# ANOVA: Yield across seasons
yield_groups = [
    df_clean[df_clean["Season"] == season]["Yield_Tonnes_Ha"]
    for season in df_clean["Season"].unique()
]

yield_f, yield_p = f_oneway(*yield_groups)

# ANOVA: Profit across seasons
profit_groups = [
    df_clean[df_clean["Season"] == season]["Profit_INR"]
    for season in df_clean["Season"].unique()
]

profit_f, profit_p = f_oneway(*profit_groups)

anova_summary = pd.DataFrame({
    "Analysis": [
        "Yield across Seasons",
        "Profit across Seasons"
    ],
    "F_Statistic": [
        yield_f,
        profit_f
    ],
    "P_Value": [
        yield_p,
        profit_p
    ],
    "Significant_at_0.05": [
        yield_p < 0.05,
        profit_p < 0.05
    ]
})

anova_summary.round(6)
```
